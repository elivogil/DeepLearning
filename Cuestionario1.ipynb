```python
import pandas as pd
import numpy as np

ds_path = "Datos/lentes.csv"

try:
    df = pd.read_csv(ds_path, encoding="cp1252", delimiter=",")
    pd.set_option('display.float_format', '{:.4f}'.format)
    print(f"✅ Dataset cargado exitosamente!")
    print(f"\n🔍 Primeras 5 filas:")
    display(df.head(5))
except Exception as e:
    print(f"❌ Error al cargar: {e}")
```

```python
df = df.drop(columns=["Id"])
df["Edad"] = df["Edad"].map({"Joven":1, "pre_presb":2, "Presbicia":3})
df["Prescripcion"] = df["Prescripcion"].map({"Miopía":1, "Hipermetropía":2})
df["Astigmatismo"] = df["Astigmatismo"].map({"NO":1, "SI":2})
df["Lagrimas"] = df["Lagrimas"].map({"Reducida":1, "Normal":2})
df["Diagnostico"] = np.where(df["Diagnostico"] == "Lentes_Blandos", 1, 0)

display(df)
```

```python
import sys
import os

# Ruta absoluta a la carpeta "Fuentes"
ruta_fuentes = os.path.abspath(os.path.join(os.getcwd(), "Fuentes"))
sys.path.append(ruta_fuentes)

# Ahora podés importar
from ClassPerceptron import Perceptron


ppn = Perceptron(alpha=0.1, n_iter=200)

np.set_printoptions(suppress=True, precision=4)

X = df.drop(columns=["Diagnostico"]).to_numpy()
T = df["Diagnostico"].to_numpy()

ppn.fit(X,T)

X = df.drop(columns=["Diagnostico"]).to_numpy()
T = df["Diagnostico"].to_numpy()

Y = ppn.predict(X)
aciertos = np.sum(Y == T)

if (aciertos == T.size):
    print("Clasifico todo correctamente")
else:
    print("Solo clasifico",aciertos,"de",T.size)

print(aciertos)
```
